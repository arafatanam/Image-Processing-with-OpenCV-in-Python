### **Reading an image**

```python
import cv2
import pandas as pd
import numpy as np
import time
from google.colab.patches import cv2_imshow
from IPython.display import clear_output

# Read image (BGR)
img = cv2.imread('/content/fruits.jpg')

# Type of image
print("Type of image: ", type(img))

# Shape of image
print("\nShape of image: ", img.shape, "\n")

# Display image
cv2_imshow(img)
```

### **Converting image to grayscale**

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(img_gray)
```

### **Using RGB color channels**

```python
# Make copies for each color removal operation
img_no_blue  = img.copy()
img_no_green = img.copy()
img_no_red   = img.copy()

# Remove individual color channels
img_no_blue[:, :, 0] = 0   # Remove Blue
img_no_green[:, :, 1] = 0  # Remove Green
img_no_red[:, :, 2] = 0    # Remove Red

# Display all images
img_all = np.hstack((img_no_blue, img_no_green, img_no_red))
cv2_imshow(img_all)
```

### **Resizing image**

```python
print("Shape of image before: ", img.shape)

# Resize the image to 256x256 pixels
img_resize = cv2.resize(img, (256, 256))
print("\nShape of image after: ", img_resize.shape)
cv2_imshow(img_resize)

# Resize the image to 50% of its original size
img_50 = cv2.resize(img, (img.shape[1]//2, img.shape[0]//2))
print("\nShape of image after 50% reduction: " ,img_50.shape)
cv2_imshow(img_50)
```

### **Flipping image**

```python
# Create flipped versions of the image
img_flip_vertical   = cv2.flip(img, 0)   # Flip vertically (top ↔ bottom)
img_flip_horizontal = cv2.flip(img, 1)   # Flip horizontally (left ↔ right)
img_flip_both       = cv2.flip(img, -1)  # Flip both vertically and horizontally

# Combine all flipped images side by side for comparison
img_all_flips = np.hstack((img_flip_vertical, img_flip_horizontal, img_flip_both))

# Display all flipped images together
cv2_imshow(img_all_flips)
```

### **Cropping image**

```python
img_crop = img[0:300, 0:300]
cv2_imshow(img_crop)
```

### **Saving image**

```python
cv2.imwrite('fruits_copy.jpg', img)
```

### **Drawing shapes in image**

```python
# Create an image
img = np.zeros((512, 512, 3))
cv2_imshow(img)
```

```python
# Draw a rectangle on the image
# Syntax: cv2.rectangle(image, top_left, bottom_right, color, thickness)
# Color format: (B, G, R)

top_left = (100, 100)
bottom_right = (300, 300)
color = (255, 0, 0)
thickness = 2 # -1 for fill

cv2.rectangle(img, top_left, bottom_right, color, thickness)
cv2_imshow(img)
```

```python
# Draw a circle on the image
# Syntax: cv2.circle(image, center_coordinates, radius, color, thickness)
# Color format: (B, G, R)

center = (400, 400)
radius = 100
color = (0, 255, 0)
thickness = 3 # -1 for fill

cv2.circle(img, center, radius, color, thickness)
cv2_imshow(img)
```

```python
# Draw a line on the image
# Syntax: cv2.line(image, start_point, end_point, color, thickness)
# Color format: (B, G, R)

start_point = (0, 0)
end_point = (512, 512)
color = (0, 0, 255)
thickness = 2

cv2.line(img, start_point, end_point, color, thickness)
cv2_imshow(img)
```

```python
# Add text to the image
# Syntax: cv2.putText(image, text, position, font, font_scale, color, thickness)
# Color format: (B, G, R)

text = "Hello World"
position = (50, 250) # Bottom-left corner of the text (x, y)
font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 1
color = (0, 255, 255)
thickness = 2

cv2.putText(img, text, position, font, font_scale, color, thickness)
cv2_imshow(img)
```
